```python
# Import necessary libraries

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter
import numpy as np
from PIL import Image, ImageDraw
```

```python
# Download required NLTK resources

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# Unique Student Hobbies Dataset

text = """
artificialintelligence artificialintelligence artificialintelligence artificialintelligence artificialintelligence
cybersecurity cybersecurity cybersecurity cybersecurity cybersecurity
cloudcomputing cloudcomputing cloudcomputing cloudcomputing
machinelearning machinelearning machinelearning machinelearning
datascience datascience datascience datascience
webdevelopment webdevelopment webdevelopment
blockchain blockchain blockchain
devops devops devops
robotics robotics robotics
cryptography cryptography cryptography
quantumcomputing quantumcomputing
supercomputing supercomputing
deeplearning deeplearning
virtualreality virtualreality
augmentedreality augmentedreality
edgecomputing edgecomputing
microservices microservices
game-development game-development
computervision computervision
naturalprocessing naturalprocessing
automation automation
fullstack fullstack
frontend frontend
backend backend
networking networking
sysadmin sysadmin
database database
algorithms algorithms
promptengineering promptengineering
ethicalhacking ethicalhacking
nanotechnology nanotechnology
iot iot
linux linux
scrum scrum
agile agile
open-source open-source
containerization containerization
kubernetes kubernetes
docker docker
biometrics biometrics
bioinformatics bioinformatics
telecommunications telecommunications
appdevelopment appdevelopment
ui-ux ui-ux
data-analytics data-analytics
software-testing software-testing
generative-ai generative-ai
neural-networks neural-networks
bigdata bigdata
microcontrollers microcontrollers
semiconductors semiconductors
embedded-systems embedded-systems
telemetry telemetry
cad-design cad-design
printed-circuit-board printed-circuit-board
"""

print(text)
```

```python
# Convert text into individual words

words = word_tokenize(text.lower())

print("Total words:", len(words))
print(words[:50])

# Remove common English stopwords

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word.isalpha() and word not in stop_words
]

print("\nFiltered words:")
print(filtered_words[:50])
```

```python
# Count the frequency of each hobby

word_freq = Counter(filtered_words)

print("Hobby Frequencies:\n")

for hobby, frequency in word_freq.most_common():
    print(f"{hobby}: {frequency}")
```

```python
# Create a full and dense Word Cloud

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white",
    colormap="viridis",
    max_words=80,
    min_font_size=8,
    max_font_size=95,
    margin=3,
    prefer_horizontal=0.9,
    relative_scaling=0.5,
    collocations=False,
    random_state=25
).generate_from_frequencies(word_freq)


# Convert WordCloud to image
wc_image = wordcloud.to_image().convert("RGB")


# Add personalized heading in the center
from PIL import ImageDraw, ImageFont

draw = ImageDraw.Draw(wc_image)

font_path = "C:/Windows/Fonts/arialbd.ttf"

try:
    heading_font = ImageFont.truetype(font_path, 30)
except:
    heading_font = ImageFont.load_default()


heading = "AARUSH HOBBIES"

# Find heading dimensions
bbox = draw.textbbox((0, 0), heading, font=heading_font)

text_width = bbox[2] - bbox[0]
text_height = bbox[3] - bbox[1]

x = (1000 - text_width) // 2
y = (500 - text_height) // 2


# Small white box behind the heading
draw.rounded_rectangle(
    (
        x - 12,
        y - 7,
        x + text_width + 12,
        y + text_height + 7
    ),
    radius=8,
    fill="white"
)


# Write heading
draw.text(
    (x, y),
    heading,
    fill="black",
    font=heading_font
)


# Display final Word Cloud
plt.figure(figsize=(15, 7))
plt.imshow(wc_image)
plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# Save the final Word Cloud as JPG

wc_image.save(
    "Aarush_Hobbies_WordCloud.jpg",
    quality=95
)

print("JPG saved successfully!")
print("File name: Aarush_Hobbies_WordCloud.jpg")
```
